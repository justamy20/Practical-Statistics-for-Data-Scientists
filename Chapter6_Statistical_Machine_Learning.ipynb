# Chapter 6: Statistical Machine Learning

## Objective
Memahami:
- Bias-Variance Tradeoff
- Overfitting & Underfitting
- Regularization (Ridge & Lasso)
- Model Complexity

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures
```

## Bias-Variance Tradeoff

- Bias: error karena model terlalu sederhana
- Variance: error karena model terlalu kompleks

Tujuan:
Menyeimbangkan bias dan variance agar model optimal

## Overfitting vs Underfitting

- Underfitting:
  Model terlalu sederhana → tidak menangkap pola

- Overfitting:
  Model terlalu kompleks → mengikuti noise

Tujuan:
Mencari model dengan generalisasi terbaik

```python
# buat data
np.random.seed(42)
X = np.linspace(0, 10, 50)
y = 3*X + np.random.normal(0, 5, 50)

X = X.reshape(-1, 1)

plt.scatter(X, y)
plt.title("Data")
plt.show()
```

```python
# model linear (underfitting)
model_simple = LinearRegression()
model_simple.fit(X, y)
y_simple = model_simple.predict(X)

# model kompleks (overfitting)
poly = PolynomialFeatures(degree=8)
X_poly = poly.fit_transform(X)

model_complex = LinearRegression()
model_complex.fit(X_poly, y)
y_complex = model_complex.predict(X_poly)

plt.scatter(X, y)
plt.plot(X, y_simple, label="Simple Model")
plt.plot(X, y_complex, label="Complex Model")
plt.legend()
plt.title("Underfitting vs Overfitting")
plt.show()
```

## Regularization

Regularization digunakan untuk mengurangi overfitting.

### Ridge Regression (L2)

$$
Loss = RSS + \lambda \sum \beta_j^2
$$

### Lasso Regression (L1)

$$
Loss = RSS + \lambda \sum |\beta_j|
$$

- λ (lambda) mengontrol kompleksitas model

```python
ridge = Ridge(alpha=1.0)
ridge.fit(X, y)
y_ridge = ridge.predict(X)

plt.scatter(X, y)
plt.plot(X, y_ridge)
plt.title("Ridge Regression")
plt.show()
```

```python
lasso = Lasso(alpha=0.5)
lasso.fit(X, y)
y_lasso = lasso.predict(X)

plt.scatter(X, y)
plt.plot(X, y_lasso)
plt.title("Lasso Regression")
plt.show()
```

## Model Complexity

- Model sederhana → bias tinggi
- Model kompleks → variance tinggi

Tujuan:
Mencari titik optimal antara bias dan variance

## Insight

- Overfitting adalah masalah utama dalam machine learning
- Model harus diuji pada data baru
- Regularization membantu mengontrol kompleksitas
- Tradeoff bias-variance sangat penting
- Model terbaik bukan yang paling kompleks, tapi yang paling generalizable
