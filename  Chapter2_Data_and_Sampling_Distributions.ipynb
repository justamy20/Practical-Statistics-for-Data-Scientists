# Chapter 2: Data and Sampling Distributions

## Objective
Memahami:
- Random sampling
- Sampling bias
- Sampling distribution
- Central Limit Theorem
- Standard Error
- Bootstrap

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Random Sampling

Random sampling adalah proses mengambil sample dari population
secara acak sehingga setiap data memiliki peluang yang sama.

Tujuan:
- Menghindari bias
- Mendapatkan sample yang representatif

```python
population = np.random.normal(50, 10, 10000)

sample = np.random.choice(population, size=100)

print("Population Mean:", np.mean(population))
print("Sample Mean:", np.mean(sample))
```

## Sampling Bias

Sampling bias terjadi ketika sample tidak mewakili population.

Contoh:
- Survey hanya dilakukan di kota besar
- Data tidak acak

Akibat:
- Estimasi menjadi tidak akurat

```python
# contoh bias (ambil hanya nilai besar)
biased_sample = population[population > 60]

print("Biased Mean:", np.mean(biased_sample))
print("True Mean:", np.mean(population))
```

## Sampling Distribution

Sampling distribution adalah distribusi dari statistik (misalnya mean)
yang diperoleh dari banyak sample.

Langkah:
1. Ambil banyak sample
2. Hitung mean tiap sample
3. Plot distribusi

```python
sample_means = []

for _ in range(1000):
    sample = np.random.choice(population, size=100)
    sample_means.append(np.mean(sample))

plt.hist(sample_means, bins=30)
plt.title("Sampling Distribution of Mean")
plt.show()
```

## Standard Error

Standard Error mengukur variasi dari sample mean.

Rumus:

$$
SE = \frac{\sigma}{\sqrt{n}}
$$

- σ = standard deviation population
- n = ukuran sample

Semakin besar n → SE semakin kecil

```python
population_std = np.std(population)
n = 100

SE = population_std / np.sqrt(n)
print("Standard Error:", SE)
```

## Central Limit Theorem (CLT)

CLT menyatakan bahwa distribusi sample mean akan mendekati normal
meskipun data asli tidak normal.

Syarat:
- Sample cukup besar (n ≥ 30)

```python
# data tidak normal
data = np.random.exponential(scale=2, size=10000)

sample_means = []

for _ in range(1000):
    sample = np.random.choice(data, size=50)
    sample_means.append(np.mean(sample))

plt.hist(sample_means, bins=30)
plt.title("CLT Demonstration")
plt.show()
```

## Bootstrap

Bootstrap adalah metode resampling:
- Mengambil sample berulang dengan replacement

Tujuan:
- Mengestimasi distribusi statistik
- Mengukur ketidakpastian

```python
bootstrap_means = []

for _ in range(1000):
    sample = np.random.choice(sample, size=100, replace=True)
    bootstrap_means.append(np.mean(sample))

plt.hist(bootstrap_means, bins=30)
plt.title("Bootstrap Distribution")
plt.show()
```

## Insight

- Random sampling sangat penting untuk menghindari bias
- Sampling bias dapat menyebabkan kesalahan besar
- Sampling distribution membantu memahami variasi estimasi
- Central Limit Theorem memungkinkan asumsi normalitas
- Bootstrap digunakan untuk estimasi tanpa asumsi distribusi
