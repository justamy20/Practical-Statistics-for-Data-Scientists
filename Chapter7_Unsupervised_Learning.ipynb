# Chapter 7: Unsupervised Learning

## Objective
Memahami:
- Clustering
- K-Means
- Distance Measure
- Principal Component Analysis (PCA)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

## Unsupervised Learning

Unsupervised learning digunakan ketika data tidak memiliki label.

Tujuan:
- Menemukan pola dalam data
- Mengelompokkan data

## Clustering

Clustering adalah teknik untuk mengelompokkan data berdasarkan kemiripan.

Contoh:
- Segmentasi pelanggan
- Pengelompokan produk

```python
# buat data cluster
np.random.seed(42)

from sklearn.datasets import make_blobs
X, _ = make_blobs(n_samples=200, centers=3, random_state=42)

plt.scatter(X[:, 0], X[:, 1])
plt.title("Data")
plt.show()
```

## K-Means Clustering

K-Means membagi data menjadi K cluster.

Langkah:
1. Tentukan jumlah cluster (K)
2. Inisialisasi centroid
3. Assign data ke centroid terdekat
4. Update centroid

Tujuan:
Meminimalkan jarak dalam cluster

## Euclidean Distance

$$
d(x, y) = \sqrt{\sum (x_i - y_i)^2}
$$

Digunakan untuk mengukur jarak antar data

```python
kmeans = KMeans(n_clusters=3)
kmeans.fit(X)

labels = kmeans.labels_
centroids = kmeans.cluster_centers_

plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(centroids[:, 0], centroids[:, 1], marker='x')
plt.title("K-Means Clustering")
plt.show()
```

## Principal Component Analysis (PCA)

PCA digunakan untuk:
- Mengurangi dimensi data
- Mempertahankan informasi penting

Tujuan:
Menyederhanakan data

## PCA Concept

PCA mencari arah dengan variansi terbesar.

Secara matematis:
- Berdasarkan eigenvector dan eigenvalue dari covariance matrix

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.scatter(X_pca[:, 0], X_pca[:, 1])
plt.title("PCA Result")
plt.show()
```

## Insight

- Unsupervised learning tidak membutuhkan label
- K-Means adalah metode clustering paling umum
- Distance menentukan hasil clustering
- PCA membantu mengurangi dimensi data
- Digunakan untuk eksplorasi data dan preprocessing
