# Chapter 5: Classification

## Objective
Memahami:
- Classification
- Logistic Regression
- Confusion Matrix
- Accuracy, Precision, Recall

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score
```

## Classification

Classification adalah teknik untuk memprediksi kategori (label).

Contoh:
- Email spam atau tidak
- Lulus atau tidak

```python
# contoh data sederhana
X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

plt.scatter(X, y)
plt.title("Data Classification")
plt.show()
```

## Logistic Regression

Digunakan untuk klasifikasi biner.

Rumus:

$$
P(Y=1) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 x)}}
$$

- Output berupa probabilitas
- Nilai antara 0 dan 1

```python
model = LogisticRegression()
model.fit(X, y)

y_pred = model.predict(X)
print("Predictions:", y_pred)
```

## Confusion Matrix

|               | Predicted Positive | Predicted Negative |
|---------------|------------------|------------------|
| Actual Positive | True Positive (TP) | False Negative (FN) |
| Actual Negative | False Positive (FP) | True Negative (TN) |

```python
cm = confusion_matrix(y, y_pred)
print("Confusion Matrix:\n", cm)
```

## Accuracy

$$
Accuracy = \frac{TP + TN}{Total}
$$

Mengukur seberapa banyak prediksi yang benar

```python
acc = accuracy_score(y, y_pred)
print("Accuracy:", acc)
```

## Precision

$$
Precision = \frac{TP}{TP + FP}
$$

## Recall

$$
Recall = \frac{TP}{TP + FN}
$$

- Precision → ketepatan prediksi positif
- Recall → kemampuan menangkap semua positif

```python
precision = precision_score(y, y_pred)
recall = recall_score(y, y_pred)

print("Precision:", precision)
print("Recall:", recall)
```

## Insight

- Classification digunakan untuk data kategori
- Logistic regression menghasilkan probabilitas
- Confusion matrix membantu evaluasi model
- Accuracy tidak selalu cukup (perlu precision & recall)
- Model harus dipilih sesuai kebutuhan masalah
