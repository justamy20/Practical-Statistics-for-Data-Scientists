# Chapter 4: Regression and Prediction

## Objective
Memahami:
- Linear Regression
- Prediction
- Residual
- Model Evaluation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## Linear Regression

Linear regression digunakan untuk memodelkan hubungan antara variabel X dan Y.

Rumus:

$$
y = \beta_0 + \beta_1 x + \epsilon
$$

- β0 = intercept
- β1 = slope
- ε = error

```python
# contoh data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 4, 5, 4, 5])

plt.scatter(X, y)
plt.title("Data")
plt.show()
```

```python
model = LinearRegression()
model.fit(X, y)

print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])
```

```python
y_pred = model.predict(X)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.title("Regression Line")
plt.show()
```

## Residual

Residual adalah selisih antara nilai aktual dan prediksi.

Rumus:

$$
e_i = y_i - \hat{y}_i
$$

Residual digunakan untuk mengevaluasi model

```python
residuals = y - y_pred
print("Residuals:", residuals)
```

## Mean Squared Error (MSE)

$$
MSE = \frac{1}{n} \sum (y_i - \hat{y}_i)^2
$$

- Semakin kecil MSE → model semakin baik

```python
mse = np.mean(residuals**2)
print("MSE:", mse)
```

## Overfitting

Overfitting terjadi ketika model terlalu kompleks dan
mengikuti noise dalam data.

Ciri:
- Training error kecil
- Testing error besar

```python
# contoh overfitting dengan polynomial
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=4)
X_poly = poly.fit_transform(X)

model_poly = LinearRegression()
model_poly.fit(X_poly, y)

y_poly_pred = model_poly.predict(X_poly)

plt.scatter(X, y)
plt.plot(X, y_poly_pred)
plt.title("Overfitting Example")
plt.show()
```

## Insight

- Linear regression memodelkan hubungan linear
- Residual menunjukkan error model
- MSE digunakan untuk evaluasi
- Model terlalu kompleks dapat menyebabkan overfitting
- Model sederhana sering lebih generalizable
