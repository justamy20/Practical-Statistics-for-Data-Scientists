# Chapter 1: Exploratory Data Analysis (EDA)

## Objective
Memahami konsep dasar eksplorasi data:
- Mean, Median
- Variance & Standard Deviation
- Percentile
- Distribusi data (Histogram & Boxplot)
- Frequency Table

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Dataset sederhana (bisa kamu ganti nanti)
data = [3, 5, 1, 2, 8, 10, 3, 6]
df = pd.DataFrame(data, columns=['Value'])

df
```

## Mean (Rata-rata)

Rumus:

$$
\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i
$$

- Mean adalah rata-rata dari seluruh data
- Sensitif terhadap outlier

```python
mean = np.mean(df['Value'])
print("Mean:", mean)
```

## Median

- Median adalah nilai tengah dari data yang sudah diurutkan
- Lebih tahan terhadap outlier dibanding mean

```python
median = np.median(df['Value'])
print("Median:", median)
```

## Variance & Standard Deviation

Variance:

$$
s^2 = \frac{\sum (x_i - \bar{x})^2}{n-1}
$$

Standard Deviation:

$$
s = \sqrt{s^2}
$$

- Variance mengukur penyebaran data
- Standard deviation lebih mudah diinterpretasi

```python
variance = np.var(df['Value'], ddof=1)
std_dev = np.std(df['Value'], ddof=1)

print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

## Percentile

- 25% → Q1
- 50% → Median
- 75% → Q3

Digunakan untuk memahami distribusi data

```python
percentiles = np.percentile(df['Value'], [25, 50, 75])
print("Percentiles (25%, 50%, 75%):", percentiles)
```

## Boxplot

- Menampilkan distribusi data
- Bisa mendeteksi outlier

Bagian:
- Box = Q1 sampai Q3
- Garis tengah = median
- Titik = outlier

```python
plt.boxplot(df['Value'])
plt.title("Boxplot")
plt.show()
```

## Histogram

- Menampilkan distribusi frekuensi data
- Data dibagi ke dalam beberapa interval (bins)

```python
plt.hist(df['Value'], bins=5)
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Frequency Table

- Mengelompokkan data ke dalam interval
- Menunjukkan jumlah data di setiap interval

```python
binned = pd.cut(df['Value'], bins=5)
print(binned.value_counts())
```

## Insight

- Mean mudah dipengaruhi outlier
- Median lebih stabil
- Variance menunjukkan sebaran data
- Boxplot membantu melihat outlier
- Histogram membantu memahami distribusi data
